```python
import matplotlib.pyplot as plt

def line_chart(ax, x, y, title='', xlabel='', ylabel='', label=None):
    ax.plot(x, y, marker='o', label=label)
    ax.set_title(title)
    ax.set_xlabel(xlabel)
    ax.set_ylabel(ylabel)
    if label:
        ax.legend()
    return ax
```
