````python
def _run_checks():
    total = 5
    passed = 0
    _DATA = [100.0, 125.0, 150.0, 250.0, 400.0, 450.0, 680.0, 1000.0, 1050.0, 1800.0]

    # Check 1: defined and callable without crashing
    try:
        assert 'histogram' in globals()
        _fig, _ax = plt.subplots()
        histogram(_ax, _DATA)
        plt.close(_fig)
        passed += 1; print('\u2705 Check 1: histogram is defined and callable')
    except Exception as e:
        print(f'\u274c Check 1: {e}')
        plt.close('all')
        print(f'\nScore: {passed}/{total}'); return

    # Check 2: histogram bars (patches) are rendered
    try:
        _fig, _ax = plt.subplots()
        histogram(_ax, _DATA, bins=5)
        assert len(_ax.patches) > 0, 'no histogram bars rendered'
        plt.close(_fig)
        passed += 1; print(f'\u2705 Check 2: {len(_ax.patches)} bar(s) rendered')
    except Exception as e:
        plt.close('all'); print(f'\u274c Check 2: {e}')

    # Check 3: bins parameter respected — more bins = more bars
    try:
        _fig4, _ax4 = plt.subplots()
        histogram(_ax4, _DATA, bins=4)
        _cnt4 = len(_ax4.patches)
        plt.close(_fig4)

        _fig8, _ax8 = plt.subplots()
        histogram(_ax8, _DATA, bins=8)
        _cnt8 = len(_ax8.patches)
        plt.close(_fig8)

        assert _cnt8 >= _cnt4, \
            f'bins=8 should give >= bars than bins=4 ({_cnt8} vs {_cnt4})'
        passed += 1; print(f'\u2705 Check 3: bins=4 gives {_cnt4} bars, bins=8 gives {_cnt8}')
    except Exception as e:
        plt.close('all'); print(f'\u274c Check 3: {e}')

    # Check 4: ylabel is 'Count', title and xlabel are set
    try:
        _fig, _ax = plt.subplots()
        histogram(_ax, _DATA, bins=5,
                  title='Revenue Dist', xlabel='Revenue ($)')
        assert _ax.get_ylabel() == 'Count', \
            f'ylabel={_ax.get_ylabel()!r}, expected "Count"'
        assert _ax.get_title() == 'Revenue Dist'
        assert _ax.get_xlabel() == 'Revenue ($)'
        plt.close(_fig)
        passed += 1; print('\u2705 Check 4: ylabel="Count", title and xlabel set')
    except Exception as e:
        plt.close('all'); print(f'\u274c Check 4: {e}')

    # Check 5: figure saves to a valid PNG
    _tmp = '/tmp/day039_ex04.png'
    try:
        _fig, _ax = plt.subplots()
        histogram(_ax, _DATA, bins=5,
                  title='Revenue Distribution', xlabel='Revenue ($)')
        _fig.savefig(_tmp, bbox_inches='tight', dpi=72)
        plt.close(_fig)
        assert os.path.exists(_tmp) and os.path.getsize(_tmp) > 1000
        with open(_tmp, 'rb') as f:
            assert f.read(4) == b'\x89PNG'
        os.remove(_tmp)
        passed += 1; print('\u2705 Check 5: figure saves as a valid PNG')
    except Exception as e:
        plt.close('all')
        if os.path.exists(_tmp): os.remove(_tmp)
        print(f'\u274c Check 5: {e}')

    if passed == total:
        print('\U0001f389 Exercise complete!')
    print(f'\nScore: {passed}/{total}')


_run_checks()
````

import matplotlib.pyplot as plt

def histogram(ax, data, bins=10, title='', xlabel=''):

    ax.set_title(title)
    ax.set_xlabel(xlabel)
    ax.set_ylabel('Count')
