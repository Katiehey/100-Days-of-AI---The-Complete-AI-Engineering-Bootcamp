````python
def _run_checks():
    total = 5
    passed = 0

    # Check 1: defined and callable without crashing
    try:
        assert 'bar_chart' in globals()
        _fig, _ax = plt.subplots()
        bar_chart(_ax, ['A', 'B'], [1, 2])
        plt.close(_fig)
        passed += 1; print('\u2705 Check 1: bar_chart is defined and callable')
    except Exception as e:
        print(f'\u274c Check 1: {e}')
        plt.close('all')
        print(f'\nScore: {passed}/{total}'); return

    # Check 2: correct number of bars
    try:
        _fig, _ax = plt.subplots()
        bar_chart(_ax, ['A', 'B', 'C'], [10, 20, 15])
        assert len(_ax.patches) == 3, \
            f'expected 3 bars, got {len(_ax.patches)}'
        plt.close(_fig)
        passed += 1; print('\u2705 Check 2: 3 bars rendered for 3 labels')
    except Exception as e:
        plt.close('all'); print(f'\u274c Check 2: {e}')

    # Check 3: title is set
    try:
        _fig, _ax = plt.subplots()
        bar_chart(_ax, PRODUCTS, REVENUES, title='Revenue')
        assert _ax.get_title() == 'Revenue', \
            f'title={_ax.get_title()!r}, expected "Revenue"'
        plt.close(_fig)
        passed += 1; print('\u2705 Check 3: title is set correctly')
    except Exception as e:
        plt.close('all'); print(f'\u274c Check 3: {e}')

    # Check 4: xlabel and ylabel are set
    try:
        _fig, _ax = plt.subplots()
        bar_chart(_ax, PRODUCTS, REVENUES,
                  title='T', xlabel='Product', ylabel='Revenue ($)')
        assert _ax.get_xlabel() == 'Product', \
            f'xlabel={_ax.get_xlabel()!r}'
        assert _ax.get_ylabel() == 'Revenue ($)', \
            f'ylabel={_ax.get_ylabel()!r}'
        plt.close(_fig)
        passed += 1; print('\u2705 Check 4: xlabel="Product", ylabel="Revenue ($)"')
    except Exception as e:
        plt.close('all'); print(f'\u274c Check 4: {e}')

    # Check 5: figure saves to a valid PNG
    _tmp = '/tmp/day039_ex01.png'
    try:
        _fig, _ax = plt.subplots()
        bar_chart(_ax, PRODUCTS, REVENUES,
                  title='Revenue by Product', xlabel='Product', ylabel='Revenue ($)')
        _fig.savefig(_tmp, bbox_inches='tight', dpi=72)
        plt.close(_fig)
        assert os.path.exists(_tmp) and os.path.getsize(_tmp) > 1000, \
            f'saved file missing or too small'
        with open(_tmp, 'rb') as f:
            assert f.read(4) == b'\x89PNG', 'not a valid PNG'
        os.remove(_tmp)
        passed += 1; print('\u2705 Check 5: figure saves as a valid PNG')
    except Exception as e:
        plt.close('all')
        if os.path.exists(_tmp): os.remove(_tmp)
        print(f'\u274c Check 5: {e}')

    if passed == total:
        print('\U0001f389 Exercise complete!')
    print(f'\nScore: {passed}/{total}')


_run_checks()
````

```python
import matplotlib.pyplot as plt

def bar_chart(ax, labels, values, title='', xlabel='', ylabel=''):
    ax.bar(labels, values)
    ax.set_title(title)
    ax.set_xlabel(xlabel)
    ax.set_ylabel(ylabel)
    ax.tick_params(axis='x', rotation=45)
    return ax
```
