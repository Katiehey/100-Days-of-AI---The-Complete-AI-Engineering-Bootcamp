## Check Your Work

````python
def _run_checks():
    total = 5
    passed = 0

    # Check 1: defined, returns a DataFrame
    try:
        assert 'load_csv' in globals()
        result = load_csv(SAMPLE_CSV)
        assert isinstance(result, pd.DataFrame), \
            f'expected DataFrame, got {type(result).__name__}'
        passed += 1; print('\u2705 Check 1: load_csv returns a DataFrame')
    except Exception as e:
        print(f'\u274c Check 1: {e}')
        print(f'\nScore: {passed}/{total}'); return

    # Check 2: correct shape
    try:
        df = load_csv(SAMPLE_CSV)
        assert df.shape == (4, 3), f'shape={df.shape}, expected (4, 3)'
        passed += 1; print('\u2705 Check 2: shape=(4, 3)')
    except Exception as e:
        print(f'\u274c Check 2: {e}')

    # Check 3: correct column names
    try:
        df = load_csv(SAMPLE_CSV)
        assert list(df.columns) == ['product', 'price', 'quantity'], \
            f'columns={list(df.columns)}'
        passed += 1; print("\u2705 Check 3: columns=['product', 'price', 'quantity']")
    except Exception as e:
        print(f'\u274c Check 3: {e}')

    # Check 4: first row values are correct
    try:
        df = load_csv(SAMPLE_CSV)
        assert df.iloc[0]['product'] == 'apple', \
            f"first product={df.iloc[0]['product']!r}, expected 'apple'"
        assert df.iloc[0]['price'] == 1.5, \
            f"first price={df.iloc[0]['price']}, expected 1.5"
        passed += 1; print("\u2705 Check 4: first row is apple, price=1.5")
    except Exception as e:
        print(f'\u274c Check 4: {e}')

    # Check 5: price and quantity are numeric
    try:
        df = load_csv(SAMPLE_CSV)
        assert pd.api.types.is_numeric_dtype(df['price']), \
            f"price dtype={df['price'].dtype}, expected numeric"
        assert pd.api.types.is_numeric_dtype(df['quantity']), \
            f"quantity dtype={df['quantity'].dtype}, expected numeric"
        passed += 1; print('\u2705 Check 5: price and quantity are numeric dtypes')
    except Exception as e:
        print(f'\u274c Check 5: {e}')

    if passed == total:
        print('\U0001f389 Exercise complete!')
    print(f'\nScore: {passed}/{total}')


_run_checks()
````
