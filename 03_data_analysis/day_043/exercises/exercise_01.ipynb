```python
def get_db_schema(conn) -> str:
    cur = conn.cursor()
    cur.execute(
        "SELECT name, sql FROM sqlite_master WHERE type='table' ORDER BY name"
    )
    rows = cur.fetchall()
    if not rows:
        return 'No tables found.'
    parts = []
    for name, ddl in rows:
        parts.append(f'Table: {name}')
        parts.append(ddl)
        parts.append('')
    return '\n'.join(parts).strip()
```
