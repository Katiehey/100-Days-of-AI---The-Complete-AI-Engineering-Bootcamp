```python
import re

def is_safe_sql(sql: str) -> bool:
    normalized = re.sub(r'--[^\n]*', '', sql)
    normalized = re.sub(r'/\*.*?\*/', '', normalized, flags=re.DOTALL)
    normalized = normalized.strip().lower()
    if not normalized.startswith('select'):
        return False
    if ';' in normalized:
        return False
    return True
```
