import re

def extract_sql(response: str) -> str:
    fence = '`' * 3
    match = re.search(fence + r'sql\s*(.*?)' + fence, response, re.DOTALL)
    if match:
        return match.group(1).strip()
    match = re.search(fence + r'\s*(.*?)' + fence, response, re.DOTALL)
    if match:
        return match.group(1).strip()
    return response.strip()
