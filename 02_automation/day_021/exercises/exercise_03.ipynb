```python
def load_json_files(directory: str) -> list[dict]:
    results = []
    for p in Path(directory).glob("*.json"):
        try:
            data = json.loads(p.read_text(encoding="utf-8"))
            if isinstance(data, dict):
                data["_source"] = p.name
                results.append(data)
        except Exception:
            pass
    return results
```
