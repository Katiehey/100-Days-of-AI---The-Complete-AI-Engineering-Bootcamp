```python
def scan_directory(directory: str, pattern: str = "*") -> list[dict]:
    result = []
    for item in Path(directory).glob(pattern):
        if item.is_file():
            result.append({
                "name": item.name,
                "path": str(item),
                "size_bytes": item.stat().st_size,
                "extension": item.suffix,
            })
    return result
```
