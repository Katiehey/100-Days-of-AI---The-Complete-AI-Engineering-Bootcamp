````python
SAMPLE_EMAIL = 'From: alice@example.com\nTo: bob@example.com\nSubject: Project Kickoff Meeting\nMessage-ID: <20260717090000.alice@example.com>\nDate: Mon, 17 Jul 2026 09:00:00 +0000\nContent-Type: text/plain; charset=utf-8\n\nHi Bob,\n\nCan we schedule a kickoff meeting for the new project?\nI am available Tuesday or Wednesday afternoon.\n\nBest,\nAlice\n'


def _run_checks():
    total = 5
    passed = 0

    # Check 1: defined
    try:
        assert 'parse_email_string' in globals()
        passed += 1; print('\u2705 Check 1: parse_email_string defined')
    except Exception as e:
        print(f'\u274c Check 1: {e}')
        print(f'\nScore: {passed}/{total}')
        return

    msg = None

    # Check 2: returns a Message object
    try:
        msg = parse_email_string(SAMPLE_EMAIL)
        assert isinstance(msg, email.message.Message), \
            f'expected Message, got {type(msg)}'
        passed += 1; print('\u2705 Check 2: returns an email.message.Message')
    except Exception as e:
        print(f'\u274c Check 2: {e}')

    # Check 3: From header correct
    try:
        assert msg is not None, 'msg is None (Check 2 failed)'
        assert msg['From'] == 'alice@example.com', \
            f"From wrong: {msg['From']!r}"
        passed += 1; print("\u2705 Check 3: From='alice@example.com'")
    except Exception as e:
        print(f'\u274c Check 3: {e}')

    # Check 4: Subject header correct
    try:
        assert msg is not None, 'msg is None'
        assert msg['Subject'] == 'Project Kickoff Meeting', \
            f"Subject wrong: {msg['Subject']!r}"
        passed += 1; print("\u2705 Check 4: Subject='Project Kickoff Meeting'")
    except Exception as e:
        print(f'\u274c Check 4: {e}')

    # Check 5: Message-ID header accessible
    try:
        assert msg is not None, 'msg is None'
        msg_id = msg.get('Message-ID')
        assert msg_id is not None, 'Message-ID header not found'
        assert 'alice@example.com' in msg_id, \
            f"Message-ID doesn't look right: {msg_id!r}"
        passed += 1; print(f'\u2705 Check 5: Message-ID accessible: {msg_id}')
    except Exception as e:
        print(f'\u274c Check 5: {e}')

    if passed == total:
        print('\U0001f389 Exercise complete!')
    print(f'\nScore: {passed}/{total}')


_run_checks()
````
