## Provided: Helper Functions

```python
def read_sheet_rows(path: str, sheet_name: str | None = None) -> list[dict]:
    wb = load_workbook(path)
    ws = wb[sheet_name] if sheet_name else wb.active
    rows = list(ws.iter_rows(values_only=True))
    if not rows:
        return []
    headers = [str(h) for h in rows[0]]
    return [dict(zip(headers, row)) for row in rows[1:]]


def write_sheet_rows(
    path: str,
    sheet_name: str,
    headers: list[str],
    rows: list[dict],
) -> None:
    wb = Workbook()
    ws = wb.active
    ws.title = sheet_name
    ws.append(headers)
    for row in rows:
        ws.append([row.get(h) for h in headers])
    wb.save(path)


def append_sheet_row(path: str, sheet_name: str, row: dict) -> None:
    wb = load_workbook(path)
    ws = wb[sheet_name]
    headers = [
        ws.cell(row=1, column=c).value
        for c in range(1, ws.max_column + 1)
    ]
    ws.append([row.get(h) for h in headers])
    wb.save(path)


def bold_header_row(path: str, sheet_name: str) -> None:
    wb = load_workbook(path)
    ws = wb[sheet_name]
    bold_font = Font(bold=True)
    for cell in ws[1]:
        cell.font = bold_font
    wb.save(path)


def ai_analyze_sheet(path: str, question: str, model: str = "llama3.2") -> str:
    rows = read_sheet_rows(path)
    data_str = json.dumps(rows[:20], indent=2)
    response = ollama.chat(
        model=model,
        messages=[
            {
                "role": "system",
                "content": (
                    "You are a data analyst. Answer questions about tabular data "
                    "concisely and precisely."
                ),
            },
            {
                "role": "user",
                "content": f"Data:\n{data_str[:3000]}\n\nQuestion: {question}",
            },
        ],
    )
    return response["message"]["content"]
```

```python
def _run_project_checks():
    total = 5
    passed = 0

    # Check 1: SpreadsheetAnalyst has all required methods
    try:
        assert 'SpreadsheetAnalyst' in globals()
        for m in ('analyze', 'write_summary', 'run'):
            assert hasattr(SpreadsheetAnalyst, m), \
                f'SpreadsheetAnalyst missing method: {m}'
        passed += 1; print('\u2705 Check 1: all methods present')
    except Exception as e:
        print(f'\u274c Check 1: {e}')

    # Check 2: analyst is an instance
    try:
        assert 'analyst' in globals()
        assert isinstance(analyst, SpreadsheetAnalyst)
        passed += 1; print('\u2705 Check 2: analyst is a SpreadsheetAnalyst')
    except Exception as e:
        print(f'\u274c Check 2: {e}')

    # Check 3: result dict has required keys
    try:
        assert 'result' in globals()
        for k in ('rows_analyzed', 'analysis', 'output'):
            assert k in result, f"result missing '{k}': {list(result)}"
        passed += 1; print('\u2705 Check 3: result has rows_analyzed/analysis/output')
    except Exception as e:
        print(f'\u274c Check 3: {e}')

    # Check 4: output file exists with two sheets
    try:
        assert 'result' in globals()
        out = result.get('output', '')
        assert os.path.exists(out), f'output file not found: {out}'
        wb = load_workbook(out)
        assert 'Data' in wb.sheetnames, \
            f"'Data' sheet missing: {wb.sheetnames}"
        assert 'Summary' in wb.sheetnames, \
            f"'Summary' sheet missing: {wb.sheetnames}"
        passed += 1; print(f'\u2705 Check 4: output has Data + Summary sheets')
    except Exception as e:
        print(f'\u274c Check 4: {e}')

    # Check 5: analysis is a non-empty string
    try:
        assert 'result' in globals()
        analysis = result.get('analysis', '')
        assert isinstance(analysis, str) and len(analysis) > 20, \
            f'analysis should be non-empty str: {analysis!r}'
        passed += 1; print(f'\u2705 Check 5: analysis is {len(analysis)} chars')
    except Exception as e:
        print(f'\u274c Check 5: {e}')

    if passed == total:
        print('\U0001f389 Project complete!')
    print(f'\nScore: {passed}/{total}')


_run_project_checks()
```
