```python
def write_sheet_rows(
    path: str,
    sheet_name: str,
    headers: list[str],
    rows: list[dict],
) -> None:
    wb = Workbook()
    ws = wb.active
    ws.title = sheet_name
    ws.append(headers)
    for row in rows:
        ws.append([row.get(h) for h in headers])
    wb.save(path)
```
