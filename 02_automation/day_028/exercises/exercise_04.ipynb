## Check Your Work

````python
import tempfile, os


def _run_checks():
    total = 5
    passed = 0

    tmp = tempfile.mktemp(suffix='.xlsx')
    try:
        # Setup: create test XLSX
        wb = Workbook()
        ws = wb.active
        ws.title = 'Report'
        ws.append(['Name', 'Value', 'Status'])
        ws.append(['Alpha', 42, 'OK'])
        ws.append(['Beta', 17, 'OK'])
        wb.save(tmp)

        # Check 1: defined
        try:
            assert 'bold_header_row' in globals()
            passed += 1; print('\u2705 Check 1: bold_header_row defined')
        except Exception as e:
            print(f'\u274c Check 1: {e}')
            print(f'\nScore: {passed}/{total}')
            return

        # Check 2: call does not raise
        try:
            bold_header_row(tmp, 'Report')
            passed += 1; print('\u2705 Check 2: bold_header_row ran without error')
        except Exception as e:
            print(f'\u274c Check 2: {e}')

        # Check 3: first row cells are bold
        try:
            wb2 = load_workbook(tmp)
            ws2 = wb2['Report']
            for cell in ws2[1]:
                assert cell.font.bold is True, \
                    f'cell {cell.coordinate} not bold: font.bold={cell.font.bold}'
            passed += 1; print('\u2705 Check 3: all header cells are bold')
        except Exception as e:
            print(f'\u274c Check 3: {e}')

        # Check 4: data row cells are NOT bold (formatting not spread)
        try:
            wb3 = load_workbook(tmp)
            ws3 = wb3['Report']
            data_cell = ws3.cell(row=2, column=1)
            assert data_cell.font.bold is not True, \
                f'data cell A2 should not be bold, got {data_cell.font.bold}'
            passed += 1; print('\u2705 Check 4: data row cells are not bold')
        except Exception as e:
            print(f'\u274c Check 4: {e}')

        # Check 5: data is preserved (values unchanged)
        try:
            wb4 = load_workbook(tmp)
            ws4 = wb4['Report']
            header = [ws4.cell(row=1, column=c).value for c in range(1, 4)]
            assert header == ['Name', 'Value', 'Status'], \
                f'header values changed: {header}'
            row2_val = ws4.cell(row=2, column=1).value
            assert row2_val == 'Alpha', \
                f'data changed: A2={row2_val!r}'
            passed += 1; print('\u2705 Check 5: cell values preserved after formatting')
        except Exception as e:
            print(f'\u274c Check 5: {e}')

    finally:
        try:
            os.unlink(tmp)
        except Exception:
            pass

    if passed == total:
        print('\U0001f389 Exercise complete!')
    print(f'\nScore: {passed}/{total}')


_run_checks()
````

```python
def bold_header_row(path: str, sheet_name: str) -> None:
    wb = load_workbook(path)
    ws = wb[sheet_name]
    bold_font = Font(bold=True)
    for cell in ws[1]:
        cell.font = bold_font
    wb.save(path)
```
