## Check Your Work

````python
class BookRecord(BaseModel):
    title: str
    price: str


def _run_checks():
    total = 5
    passed = 0

    # Check 1: defined
    try:
        assert 'validate_extracted' in globals()
        passed += 1; print('\u2705 Check 1: validate_extracted defined')
    except Exception as e:
        print(f'\u274c Check 1: {e}')
        print(f'\nScore: {passed}/{total}')
        return

    valid_result = None

    # Check 2: valid dict → model instance
    try:
        valid_result = validate_extracted(
            {'title': 'Python Cookbook', 'price': '\u00a312.99'},
            BookRecord,
        )
        assert isinstance(valid_result, BookRecord), \
            f'expected BookRecord, got {type(valid_result)}'
        passed += 1; print('\u2705 Check 2: valid dict returns BookRecord instance')
    except Exception as e:
        print(f'\u274c Check 2: {e}')

    # Check 3: field values are correct
    try:
        assert valid_result is not None, 'valid_result is None (Check 2 failed)'
        assert valid_result.title == 'Python Cookbook', \
            f"expected title='Python Cookbook', got {valid_result.title!r}"
        assert valid_result.price == '\u00a312.99', \
            f"expected price='\u00a312.99', got {valid_result.price!r}"
        passed += 1; print('\u2705 Check 3: field values are correct')
    except Exception as e:
        print(f'\u274c Check 3: {e}')

    # Check 4: missing required field → None
    try:
        bad_result = validate_extracted({'title': 'Only Title'}, BookRecord)
        assert bad_result is None, \
            f'expected None for missing required field, got {bad_result!r}'
        passed += 1; print('\u2705 Check 4: missing required field returns None')
    except Exception as e:
        print(f'\u274c Check 4: {e}')

    # Check 5: extra fields in dict are ignored (Pydantic v2 default)
    try:
        with_extras = {
            'title': 'A Book', 'price': '\u00a35.00',
            'color': 'blue', 'rating': 'Four',
        }
        extra_result = validate_extracted(with_extras, BookRecord)
        assert isinstance(extra_result, BookRecord), \
            f'extra fields should be ignored, got {type(extra_result)}'
        assert extra_result.title == 'A Book', \
            f"expected title='A Book', got {extra_result.title!r}"
        passed += 1; print('\u2705 Check 5: extra fields in dict are ignored')
    except Exception as e:
        print(f'\u274c Check 5: {e}')

    if passed == total:
        print('\U0001f389 Exercise complete!')
    print(f'\nScore: {passed}/{total}')


_run_checks()
````
