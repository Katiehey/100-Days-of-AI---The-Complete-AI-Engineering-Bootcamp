## Check Your Work

````python
import io
import sys
from argparse import ArgumentParser

def _run_checks():
    total = 5
    passed = 0

    # Check 1: defined and returns ArgumentParser
    try:
        assert 'make_parser' in globals()
        parser = make_parser()
        assert isinstance(parser, ArgumentParser), \
            f'expected ArgumentParser, got {type(parser).__name__}'
        passed += 1; print('\u2705 Check 1: make_parser returns ArgumentParser')
    except Exception as e:
        print(f'\u274c Check 1: {e}')
        print(f'\nScore: {passed}/{total}')
        return

    # Check 2: parse valid args — defaults applied
    try:
        ns = parser.parse_args(['--prompt', 'hello'])
        assert ns.prompt == 'hello',      f'prompt: {ns.prompt!r}'
        assert ns.model  == 'llama3.2',   f'model default: {ns.model!r}'
        assert ns.verbose == False,        f'verbose default: {ns.verbose!r}'
        passed += 1; print('\u2705 Check 2: parse --prompt hello gives correct defaults')
    except Exception as e:
        print(f'\u274c Check 2: {e}')

    # Check 3: short flags (-p, -v) work
    try:
        ns = parser.parse_args(['-p', 'test', '-v'])
        assert ns.prompt  == 'test', f'-p: {ns.prompt!r}'
        assert ns.verbose == True,   f'-v: {ns.verbose!r}'
        passed += 1; print('\u2705 Check 3: short flags -p and -v work')
    except Exception as e:
        print(f'\u274c Check 3: {e}')

    # Check 4: --model overrides default
    try:
        ns = parser.parse_args(['--prompt', 'hi', '--model', 'llama3.1'])
        assert ns.model == 'llama3.1', f'custom model: {ns.model!r}'
        passed += 1; print('\u2705 Check 4: --model overrides default')
    except Exception as e:
        print(f'\u274c Check 4: {e}')

    # Check 5: missing --prompt raises SystemExit(2)
    try:
        old_err = sys.stderr; sys.stderr = io.StringIO()
        raised = False
        try:
            parser.parse_args([])
        except SystemExit as e:
            raised = True
            assert e.code == 2, f'expected exit code 2, got {e.code}'
        finally:
            sys.stderr = old_err
        assert raised, 'missing --prompt should raise SystemExit'
        passed += 1; print('\u2705 Check 5: missing --prompt raises SystemExit(2)')
    except Exception as e:
        print(f'\u274c Check 5: {e}')

    if passed == total:
        print('\U0001f389 Exercise complete!')
    print(f'\nScore: {passed}/{total}')


_run_checks()
````

from argparse import ArgumentParser

def make_parser() -> ArgumentParser:
    parser = ArgumentParser(
        prog='ai-tool',
        description='AI command-line tool powered by local LLM',
    )
    parser.add_argument(
        '--prompt', '-p', type=str, required=True,
        help='Prompt to send to the model',
    )
    parser.add_argument(
        '--model', '-m', type=str, default='llama3.2',
        help='Ollama model name (default: llama3.2)',
    )
    parser.add_argument(
        '--verbose', '-v', action='store_true',
        help='Print extra diagnostic output',
    )
    return parser
