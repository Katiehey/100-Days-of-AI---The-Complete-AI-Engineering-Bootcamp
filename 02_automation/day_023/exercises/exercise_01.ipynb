## Check Your Work

````python
HTML = '<!DOCTYPE html>\n<html>\n<head><title>Book Reviews</title></head>\n<body>\n  <h1>Top Books</h1>\n  <p class="intro">Curated reading list.</p>\n  <ul>\n    <li><a href="/books/python">Learn Python</a></li>\n    <li><a href="/books/ml">Machine Learning</a></li>\n    <li><a href="/books/data">Data Science</a></li>\n    <li><a>No Link</a></li>\n  </ul>\n  <h2>Fiction</h2>\n  <h2>Non-Fiction</h2>\n  <p class="highlight">Great content here.</p>\n  <p class="highlight">More highlights.</p>\n</body>\n</html>'


def _run_checks():
    total = 5
    passed = 0

    # Check 1: defined
    try:
        assert 'parse_html' in globals()
        passed += 1; print('\u2705 Check 1: parse_html defined')
    except Exception as e:
        print(f'\u274c Check 1: {e}')
        print(f'\nScore: {passed}/{total}')
        return

    soup = None

    # Check 2: returns a BeautifulSoup
    try:
        soup = parse_html(HTML)
        assert isinstance(soup, BeautifulSoup), \
            f'expected BeautifulSoup, got {type(soup)}'
        passed += 1; print('\u2705 Check 2: returns a BeautifulSoup')
    except Exception as e:
        print(f'\u274c Check 2: {e}')

    # Check 3: can find h1
    try:
        assert soup is not None, 'soup is None (Check 2 failed)'
        h1 = soup.find('h1')
        assert h1 is not None, 'soup.find("h1") returned None'
        assert h1.get_text(strip=True) == 'Top Books', \
            f"h1 text should be 'Top Books', got {h1.get_text(strip=True)!r}"
        passed += 1; print('\u2705 Check 3: soup.find("h1") returns Top Books')
    except Exception as e:
        print(f'\u274c Check 3: {e}')

    # Check 4: title element found
    try:
        assert soup is not None, 'soup is None'
        title = soup.find('title')
        assert title is not None, 'soup.find("title") returned None'
        assert title.get_text(strip=True) == 'Book Reviews', \
            f"title should be 'Book Reviews', got {title.get_text(strip=True)!r}"
        passed += 1; print('\u2705 Check 4: soup.find("title") returns Book Reviews')
    except Exception as e:
        print(f'\u274c Check 4: {e}')

    # Check 5: find_all returns multiple elements
    try:
        assert soup is not None, 'soup is None'
        ps = soup.find_all('p')
        assert len(ps) >= 3, f'expected >= 3 p elements, got {len(ps)}'
        passed += 1; print(f'\u2705 Check 5: find_all("p") returns {len(ps)} elements')
    except Exception as e:
        print(f'\u274c Check 5: {e}')

    if passed == total:
        print('\U0001f389 Exercise complete!')
    print(f'\nScore: {passed}/{total}')


_run_checks()
````
