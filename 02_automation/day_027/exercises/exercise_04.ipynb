def record_run(
    log: list[dict],
    name: str,
    result: str,
    status: str = "ok",
) -> list[dict]:
    new_record = {
        "name": name,
        "ran_at": datetime.now().isoformat(),
        "result": result,
        "status": status,
    }
    return log + [new_record]
