````python
import tempfile, os


def _run_checks():
    total = 5
    passed = 0

    tmp = tempfile.mktemp(suffix='.json')
    RECORDS = [
        {'name': 'briefing', 'ran_at': '2026-07-17T08:00:00', 'result': 'ok', 'status': 'ok'},
        {'name': 'briefing', 'ran_at': '2026-07-17T09:00:00', 'result': 'ok', 'status': 'ok'},
    ]

    # Check 1: both functions defined
    try:
        assert 'save_run_log' in globals(), 'save_run_log not defined'
        assert 'load_run_log' in globals(), 'load_run_log not defined'
        passed += 1; print('\u2705 Check 1: save_run_log and load_run_log defined')
    except Exception as e:
        print(f'\u274c Check 1: {e}')
        print(f'\nScore: {passed}/{total}')
        return

    # Check 2: load_run_log returns [] for missing file
    try:
        missing = tempfile.mktemp(suffix='.json')
        result = load_run_log(missing)
        assert result == [], f'expected [] for missing file, got {result}'
        passed += 1; print('\u2705 Check 2: missing file → []')
    except Exception as e:
        print(f'\u274c Check 2: {e}')

    # Check 3: round-trip save then load
    loaded = None
    try:
        save_run_log(tmp, RECORDS)
        loaded = load_run_log(tmp)
        assert loaded == RECORDS, f'round-trip failed: {loaded}'
        passed += 1; print(f'\u2705 Check 3: round-trip: {len(loaded)} records')
    except Exception as e:
        print(f'\u274c Check 3: {e}')

    # Check 4: save overwrites (not appends)
    try:
        single = [{'name': 'x', 'ran_at': '2026-07-17T10:00:00', 'result': 'y', 'status': 'ok'}]
        save_run_log(tmp, single)
        reloaded = load_run_log(tmp)
        assert len(reloaded) == 1, \
            f'expected 1 record after overwrite, got {len(reloaded)}'
        passed += 1; print('\u2705 Check 4: save overwrites (not appends)')
    except Exception as e:
        print(f'\u274c Check 4: {e}')

    # Check 5: empty list round-trips
    try:
        save_run_log(tmp, [])
        empty = load_run_log(tmp)
        assert empty == [], f'expected [] for empty save, got {empty}'
        passed += 1; print('\u2705 Check 5: empty list round-trips correctly')
    except Exception as e:
        print(f'\u274c Check 5: {e}')

    try:
        os.unlink(tmp)
    except Exception:
        pass

    if passed == total:
        print('\U0001f389 Exercise complete!')
    print(f'\nScore: {passed}/{total}')


_run_checks()
````

def save_run_log(path: str, records: list[dict]) -> None:
    Path(path).write_text(json.dumps(records, indent=2), encoding="utf-8")


def load_run_log(path: str) -> list[dict]:
    p = Path(path)
    if not p.exists():
        return []
    return json.loads(p.read_text(encoding="utf-8"))
