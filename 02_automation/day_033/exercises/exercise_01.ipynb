import asyncio
import ollama

async def async_chat(prompt: str, model: str = "llama3.2") -> str:
    client = ollama.AsyncClient()
    response = await client.chat(
        model=model,
        messages=[{"role": "user", "content": prompt}],
    )
    return response["message"]["content"]
