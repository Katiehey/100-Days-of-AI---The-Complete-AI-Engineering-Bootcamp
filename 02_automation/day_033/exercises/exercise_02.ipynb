## Check Your Work

````python
import asyncio
import time

async def _run_checks():
    total = 5
    passed = 0

    async def _slow(x, delay: float = 0.05):
        await asyncio.sleep(delay)
        return x

    # Check 1: defined
    try:
        assert 'gather_results' in globals()
        passed += 1; print('\u2705 Check 1: gather_results defined')
    except Exception as e:
        print(f'\u274c Check 1: {e}')
        print(f'\nScore: {passed}/{total}')
        return

    # Check 2: returns a list
    try:
        result = await gather_results([_slow(1), _slow(2), _slow(3)])
        assert isinstance(result, list), \
            f'expected list, got {type(result).__name__}'
        assert len(result) == 3, \
            f'expected 3 results, got {len(result)}'
        passed += 1; print('\u2705 Check 2: returns a list of correct length')
    except Exception as e:
        print(f'\u274c Check 2: {e}')

    # Check 3: order is preserved (coroutines finish at same time)
    try:
        result = await gather_results([_slow(30), _slow(10), _slow(20)])
        assert result == [30, 10, 20], \
            f'order not preserved: expected [30,10,20], got {result}'
        passed += 1; print('\u2705 Check 3: results in input order (not completion order)')
    except Exception as e:
        print(f'\u274c Check 3: {e}')

    # Check 4: empty list → []
    try:
        result = await gather_results([])
        assert result == [], \
            f'empty coros: expected [], got {result}'
        passed += 1; print('\u2705 Check 4: empty list → []')
    except Exception as e:
        print(f'\u274c Check 4: {e}')

    # Check 5: concurrent — 5 × 0.05s tasks finish faster than 0.20s
    try:
        start = time.time()
        coros = [_slow(i, delay=0.05) for i in range(5)]
        await gather_results(coros)
        elapsed = time.time() - start
        assert elapsed < 0.20, \
            f'5 × 0.05s concurrent should take < 0.20s, took {elapsed:.3f}s — serial?'
        passed += 1; print(f'\u2705 Check 5: concurrent execution ({elapsed:.3f}s < 0.20s)')
    except Exception as e:
        print(f'\u274c Check 5: {e}')

    if passed == total:
        print('\U0001f389 Exercise complete!')
    print(f'\nScore: {passed}/{total}')


await _run_checks()
````
