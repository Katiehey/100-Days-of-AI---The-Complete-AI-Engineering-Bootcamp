## Check Your Work

````python
from reportlab.pdfgen import canvas as _rl_canvas
from reportlab.lib.pagesizes import letter as _letter
import tempfile, os


def _run_checks():
    total = 5
    passed = 0

    # Create a 2-page test PDF
    tmp = tempfile.mktemp(suffix='.pdf')
    _c = _rl_canvas.Canvas(tmp, pagesize=_letter)
    _c.setFont('Helvetica', 12)
    _c.drawString(72, 750, 'Test Page One Content')
    _c.showPage()
    _c.drawString(72, 750, 'Test Page Two Content')
    _c.save()

    pages = None

    # Check 1: defined
    try:
        assert 'read_pdf_pages' in globals()
        passed += 1; print('\u2705 Check 1: read_pdf_pages defined')
    except Exception as e:
        print(f'\u274c Check 1: {e}')
        os.unlink(tmp)
        print(f'\nScore: {passed}/{total}')
        return

    # Check 2: returns a list
    try:
        pages = read_pdf_pages(tmp)
        assert isinstance(pages, list), f'expected list, got {type(pages)}'
        passed += 1; print('\u2705 Check 2: returns a list')
    except Exception as e:
        print(f'\u274c Check 2: {e}')

    # Check 3: list has 2 items (2-page PDF)
    try:
        assert pages is not None, 'pages is None (Check 2 failed)'
        assert len(pages) == 2, f'expected 2 pages, got {len(pages)}'
        passed += 1; print(f'\u2705 Check 3: {len(pages)} pages extracted')
    except Exception as e:
        print(f'\u274c Check 3: {e}')

    # Check 4: all items are strings
    try:
        assert pages is not None, 'pages is None'
        bad = [p for p in pages if not isinstance(p, str)]
        assert not bad, f'non-string pages: {bad}'
        passed += 1; print('\u2705 Check 4: all page items are strings')
    except Exception as e:
        print(f'\u274c Check 4: {e}')

    # Check 5: at least one page contains expected text
    try:
        assert pages is not None, 'pages is None'
        combined = ' '.join(pages)
        assert 'Test Page' in combined, \
            f"'Test Page' not found in extracted text: {combined!r}"
        passed += 1; print('\u2705 Check 5: extracted text contains expected content')
    except Exception as e:
        print(f'\u274c Check 5: {e}')

    try:
        os.unlink(tmp)
    except Exception:
        pass

    if passed == total:
        print('\U0001f389 Exercise complete!')
    print(f'\nScore: {passed}/{total}')


_run_checks()
````
