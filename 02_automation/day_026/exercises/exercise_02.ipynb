````python
from docx import Document as _Document
import tempfile, os


def _run_checks():
    total = 5
    passed = 0

    # Create a test DOCX
    tmp = tempfile.mktemp(suffix='.docx')
    _doc = _Document()
    _doc.add_heading('Meeting Summary', level=0)
    _doc.add_paragraph('The team discussed the quarterly results.')
    _doc.add_paragraph('Action items were assigned to each member.')
    _doc.save(tmp)

    text = None

    # Check 1: defined
    try:
        assert 'read_docx_text' in globals()
        passed += 1; print('\u2705 Check 1: read_docx_text defined')
    except Exception as e:
        print(f'\u274c Check 1: {e}')
        os.unlink(tmp)
        print(f'\nScore: {passed}/{total}')
        return

    # Check 2: returns a string
    try:
        text = read_docx_text(tmp)
        assert isinstance(text, str), f'expected str, got {type(text)}'
        passed += 1; print('\u2705 Check 2: returns a string')
    except Exception as e:
        print(f'\u274c Check 2: {e}')

    # Check 3: contains the heading text
    try:
        assert text is not None, 'text is None (Check 2 failed)'
        assert 'Meeting Summary' in text, \
            f"'Meeting Summary' not in text: {text!r}"
        passed += 1; print("\u2705 Check 3: contains heading 'Meeting Summary'")
    except Exception as e:
        print(f'\u274c Check 3: {e}')

    # Check 4: contains paragraph body text
    try:
        assert text is not None, 'text is None'
        assert 'quarterly results' in text, \
            f"'quarterly results' not in text: {text!r}"
        passed += 1; print("\u2705 Check 4: contains paragraph body text")
    except Exception as e:
        print(f'\u274c Check 4: {e}')

    # Check 5: multiple paragraphs are separated (newline present)
    try:
        assert text is not None, 'text is None'
        assert '\n' in text, \
            f'expected newlines between paragraphs, got {text!r}'
        passed += 1; print('\u2705 Check 5: paragraphs separated by newlines')
    except Exception as e:
        print(f'\u274c Check 5: {e}')

    try:
        os.unlink(tmp)
    except Exception:
        pass

    if passed == total:
        print('\U0001f389 Exercise complete!')
    print(f'\nScore: {passed}/{total}')


_run_checks()
````
