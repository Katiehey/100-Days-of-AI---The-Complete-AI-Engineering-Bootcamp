# Day 082 — Project: An Agent That Remembers You

## Objective

Build `memory_agent.py` — an assistant that remembers you within a conversation *and* across sessions.

## Deliverable

`memory_agent.py` with:

- `WorkingMemory` — bounded short-term turn log (`add/turns/render/clear/__len__`)
- `LongTermMemory` — durable SQLite key/value store (`remember/recall/search/all/forget/__len__/close`)
- `build_extraction_prompt(message)` / `extract_memories(message, llm_fn=None)`
- `build_memory_prompt(message, working, longterm)`
- `MemoryAgent(longterm=None, llm_fn=None, max_turns=10)` with `chat/remember/recall/profile/end_session`

## Usage (with Ollama running + llama3.2 pulled)

```python
from memory_agent import MemoryAgent, LongTermMemory
mem = LongTermMemory('user.db')          # persists across runs
agent = MemoryAgent(longterm=mem)
agent.chat("Hi, I'm Kutlwano and I love SQL.")
print(agent.chat('What do you know about me?'))
```

**The deliverable:** you run it, tell it something about yourself, end the session — and a fresh agent on the same `LongTermMemory` still knows you. Short-term memory keeps the conversation coherent; long-term memory carries your profile across sessions.
