## Objective

Build `react_agent.py` — an agent that reasons out loud in the `Thought → Action → Observation` ReAct loop before answering.

## Deliverable

`react_agent.py` with:

- `DEFAULT_TOOLS` — calculator + lookup (tools reused from Day 79)
- `parse_react_step(text) -> dict` (never raises)
- `format_step(step)` / `format_observation(result)`
- `build_react_prompt(task, tools, scratchpad) -> list[dict]`
- `execute_action(step, tools) -> str`
- `call_llm(messages, llm_fn=None) -> str`
- `run_react_agent(task, tools=None, llm_fn=None, max_iterations=10) -> dict`
- `ReactAgent(tools=None, llm_fn=None, max_iterations=10)` with `add_tool/run/history/clear_history`

## Usage (with Ollama running + llama3.2 pulled)

```python
from react_agent import ReactAgent
agent = ReactAgent()
result = agent.run('What is the speed of light in km/s, rounded to millions?')
print(result['answer'])
for step in result['trace']:
    print(step)
```

**The deliverable:** you run it, and `result['trace']` shows the agent reasoning — Thought, Action, Observation — on its way to the answer. That visible reasoning is what ReAct buys you.
