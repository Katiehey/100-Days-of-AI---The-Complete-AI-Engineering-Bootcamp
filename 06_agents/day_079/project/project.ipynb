# Day 079 — Project: A Minimal Agent from Scratch

## Objective

Build `simple_agent.py` — an agent that is an LLM wrapped in a loop, able to call tools and decide when it is finished.

## Deliverable

`simple_agent.py` with:

- `safe_calculate(expression)` — arithmetic via `ast`, never `eval`
- `DEFAULT_TOOLS` — registry of `calculator` + `word_count`
- `build_tool_descriptions(tools) -> str`
- `safe_parse_json(text) -> dict | None`
- `parse_action(text) -> dict` (never raises)
- `execute_tool(action, tools) -> str`
- `call_llm(messages, llm_fn=None) -> str`
- `build_agent_prompt(task, tools, history) -> list[dict]`
- `run_agent(task, tools=None, llm_fn=None, max_iterations=10) -> dict`
- `SimpleAgent(tools=None, llm_fn=None, max_iterations=10)` with `add_tool/run/history/clear_history`

## Usage (with Ollama running + llama3.2 pulled)

```python
from simple_agent import SimpleAgent
agent = SimpleAgent()
result = agent.run('What is 17 * 23, and how many words are in this sentence?')
print(result['answer'])
```

**The deliverable:** you run it, it loops — calling the calculator, reading the result, and finishing with an answer. That loop is the agent.
