# Day 083 — Project: A Planning Agent

## Objective

Build `planner_agent.py` — an agent that breaks a complex goal into an ordered list of subtasks and executes them step by step.

## Deliverable

`planner_agent.py` with:

- `safe_parse_list(text)` — list-aware tolerant JSON parser
- `Task` dataclass — `id, title, description, depends_on, status, result`
- `build_plan_prompt(goal, context=None)` / `parse_plan(text)`
- `topo_sort(tasks)` — Kahn's topological sort, cycle-tolerant
- `build_execution_context(task, results)` / `execute_task(...)`
- `run_plan(goal, executor_fn=None, llm_fn=None, max_tasks=20)`
- `PlannerAgent(executor_fn=None, llm_fn=None, max_tasks=20)` with `plan / execute / run / history / clear_history`

## Usage (with Ollama running + llama3.2 pulled)

```python
from planner_agent import PlannerAgent
agent = PlannerAgent()
result = agent.run('Write a short blog post about topological sort')
for task in result['tasks']:
    print(f'{task.id}: {task.title}')
print(result['answer'])
```

**The deliverable:** you give the agent a complex goal, it returns an ordered task list with each result filled in — and the final answer is the last task's output. Inspect the plan before it runs; edit it if you want. That separation is what makes a planner agent trustworthy.
