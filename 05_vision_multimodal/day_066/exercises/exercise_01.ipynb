# Day 066 — Exercise 1: Get Image Info

**What you'll build:** `get_image_info(img)` — inspect a PIL Image and return its key metadata as a dict.

**Why it matters:** Before processing any image, you need to know what you're working with. Size, mode, and channel count determine every downstream decision — how to resize, whether to convert mode, how much memory to expect.

````python
from PIL import Image
import numpy as np

# Test images for the checks
_rgb  = Image.new('RGB',  (200, 100), color=(255, 128, 0))
_rgba = Image.new('RGBA', (50, 50),   color=(0, 0, 0, 128))
_gray = Image.new('L',    (64, 32),   color=128)
````

## Task

Implement `get_image_info(img) -> dict` with keys:

| Key | Type | Description |
|-----|------|-------------|
| `width` | int | image width in pixels |
| `height` | int | image height in pixels |
| `mode` | str | color mode string |
| `num_channels` | int | number of channels |

For `num_channels`: use lookup `{'RGB': 3, 'RGBA': 4, 'L': 1}`, fall back to `len(img.getbands())`.

## Your Implementation

````python
def get_image_info(img: Image.Image) -> dict:
    """Return metadata about a PIL Image.

    Returns a dict with keys:
        width:        int — image width in pixels
        height:       int — image height in pixels
        mode:         str — color mode ('RGB', 'L', 'RGBA', etc.)
        num_channels: int — number of channels

    For num_channels, use the lookup RGB=3, RGBA=4, L=1.
    For any other mode, use len(img.getbands()).
    """
    raise NotImplementedError
````

````python
def get_image_info(img: Image.Image) -> dict:
    w, h = img.size
    channel_map = {'RGB': 3, 'RGBA': 4, 'L': 1}
    n = channel_map.get(img.mode, len(img.getbands()))
    return {'width': w, 'height': h, 'mode': img.mode, 'num_channels': n}
````

## Automated checks

## Solution

```python
def get_image_info(img: Image.Image) -> dict:
    w, h = img.size
    channel_map = {'RGB': 3, 'RGBA': 4, 'L': 1}
    n = channel_map.get(img.mode, len(img.getbands()))
    return {'width': w, 'height': h, 'mode': img.mode, 'num_channels': n}
```

**Why a lookup dict for channels?** `img.getbands()` works for all modes but the lookup is faster and more explicit for the three modes you will encounter 99% of the time. The fallback handles uncommon modes (like 'HSV' or 'LAB') gracefully without extra branches.
