# Day 078 — Project: Multimodal Media Studio

## Objective

Build `media_studio.py` — a Media Studio that combines vision description, audio transcription, and text-to-speech synthesis into a unified interface.

## Deliverable

`media_studio.py` with:

- `MEDIA_EXTENSIONS`: dict of type → extension set
- `detect_media_type(path) -> str`
- `describe_media(source, describe_fn=None) -> str`
- `transcribe_media(source, transcribe_fn=None) -> dict`
- `synthesize_speech(text, ..., tts_fn=None) -> bytes`
- `narrate_image(source, ..., describe_fn=None, tts_fn=None) -> dict`
- `process_media(source, ...) -> dict`
- `MediaStudio(describe_fn=None, transcribe_fn=None, tts_fn=None)` with `describe/transcribe/speak/narrate/process/batch`

## Usage (with Ollama + Whisper + edge-tts installed)

```python
from media_studio import MediaStudio
studio = MediaStudio()
desc = studio.describe('photo.png')
audio = studio.speak(desc)
Path('narration.mp3').write_bytes(audio)
```
