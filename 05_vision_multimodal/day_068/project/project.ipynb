## What You're Building

`document_reader.py` — a `DocumentReader` class for extracting text and structured data from images and PDFs.

**Deliverable:** A class combining Tesseract OCR (images) and pypdf (PDFs) with a clean mock-injection interface for testing.

## Setup

```bash
brew install tesseract      # OCR binary
pip install pytesseract pypdf  # already installed
```

## Design

```
DocumentReader
  .__init__(ocr_fn=None)
  .read_image(img, preprocess=True) → {text, numbers, chars, preprocess}
  .read_pdf(pdf_bytes)              → {pages, total_chars, page_count}
  .read(source, source_type)        → dispatches above
```

## Usage (with real Tesseract)

```python
from document_reader import DocumentReader
from PIL import Image

dr = DocumentReader()
img = Image.open('receipt.png')
result = dr.read_image(img)
print(result['text'])
print('Total:', max(result['numbers']))
```
