# Day 067 — Project: Vision Analyzer

## What You're Building

`vision_analyzer.py` — a `VisionAnalyzer` class backed by Ollama llava.

**Deliverable:** A class that accepts any PIL Image and returns text descriptions, extracted text, zero-shot classifications, and multi-task analysis results. Works offline with a `describe_fn` mock; works with a real image when Ollama is running.

## Setup

```bash
# Pull the vision model (one-time)
ollama pull llava
# or: ollama pull llama3.2-vision
```

## Design

```
VisionAnalyzer
  .__init__(model='llava', describe_fn=None)
  .describe(img, prompt=...)     → str
  .extract_text(img)             → str
  .classify(img, labels)         → str  (one of labels)
  .analyze(img, tasks=['...'])   → dict  (task → result)
```

## Usage (with real Ollama)

```python
from vision_analyzer import VisionAnalyzer
from PIL import Image

va = VisionAnalyzer(model='llava')
img = Image.open('your_photo.jpg')
print(va.describe(img))
print(va.classify(img, ['indoor', 'outdoor', 'food']))
print(va.analyze(img, tasks=['describe', 'objects']))
```
