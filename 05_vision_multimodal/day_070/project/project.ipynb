`image_generator.py` — an `ImageGenerator` class for text-to-image generation.

**Deliverable:** A class and utility functions for building prompts, applying style templates, generating images (with mock injection for offline testing), and stitching variation batches into comparison grids.

## Design

```
build_prompt(subject, style, quality_tags, negative_tags) -> dict
apply_style_template(base_prompt, template_name) -> dict
generate_image(prompt, negative, generate_fn, ...) -> Image.Image
generate_variations(base_prompt, n, generate_fn) -> list[Image.Image]
create_image_grid(images, cols) -> Image.Image

ImageGenerator(model_id, generate_fn)
  .generate(prompt, ...) -> Image.Image
  .batch(prompts, ...) -> list[Image.Image]
  .grid(prompts, cols, ...) -> Image.Image
```

## Usage (offline testing)

```python
from PIL import Image
from image_generator import ImageGenerator, build_prompt, apply_style_template

mock = lambda p, **kw: Image.new('RGB', (kw.get('width', 512), kw.get('height', 512)), 'steelblue')
gen = ImageGenerator(generate_fn=mock)
p = apply_style_template('a mountain lake at dawn', 'cinematic')
img = gen.generate(p['positive'], negative=p['negative'])
print(img.size)
```
