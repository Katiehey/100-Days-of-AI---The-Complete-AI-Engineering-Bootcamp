```python
async def generate_audio(text: str, voice: str, output_path: str) -> None:
    communicate = edge_tts.Communicate(text, voice)
    await communicate.save(output_path)
```

**Why this works:**
- `edge_tts.Communicate(text, voice)` creates a TTS session with the text and voice you specify.
- `.save(output_path)` sends the request to Microsoft's servers, receives the audio stream, and writes it to disk as an MP3.
- The function is `async` because the network call is I/O-bound — using async allows other work to happen while we wait for the response.
