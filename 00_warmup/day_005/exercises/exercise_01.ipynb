```python
import logging

logging.basicConfig(
    level=logging.DEBUG,
    format="%(asctime)s %(levelname)-8s %(message)s",
)

logger = logging.getLogger(__name__)

def parse_and_log(text: str) -> float:
    logger.debug("parse_and_log() | text=%r", text)
    value = float(text.strip())
    logger.info("parse_and_log() | returning %f", value)
    return value
```

**Key details:** (1) `basicConfig` is called once at module level. (2) `getLogger(__name__)`
gives the logger the module's name — log lines show exactly which file they came from.
(3) Use `%r` (repr) for the input so hidden whitespace shows up in the log.
(4) Log at DEBUG for entry (diagnostic), INFO for success (confirmation).
