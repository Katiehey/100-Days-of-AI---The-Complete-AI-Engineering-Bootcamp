## Check Your Work

Run the cell below — it calls your test functions and shows ✅ / ❌ for each.

````python
_PASS, _FAIL = '✅', '❌'

def _run_checks():
    score, total = 0, 4

    # Check 1: all three test functions exist
    required = ['test_strips_whitespace', 'test_lowercases', 'test_empty_string']
    try:
        for name in required:
            assert name in globals(), f'{name} is not defined'
        print(f'{_PASS} Check 1/{total}: all three test functions are defined')
        score += 1
    except AssertionError as e:
        print(f'{_FAIL} Check 1/{total}: {e}')
        return

    # Check 2: test_strips_whitespace passes
    try:
        test_strips_whitespace()
        print(f'{_PASS} Check 2/{total}: test_strips_whitespace() passes')
        score += 1
    except AssertionError as e:
        print(f'{_FAIL} Check 2/{total}: test_strips_whitespace() failed — {e}')
    except Exception as e:
        print(f'{_FAIL} Check 2/{total}: unexpected error — {e}')

    # Check 3: test_lowercases passes
    try:
        test_lowercases()
        print(f'{_PASS} Check 3/{total}: test_lowercases() passes')
        score += 1
    except AssertionError as e:
        print(f'{_FAIL} Check 3/{total}: test_lowercases() failed — {e}')
    except Exception as e:
        print(f'{_FAIL} Check 3/{total}: unexpected error — {e}')

    # Check 4: test_empty_string passes
    try:
        test_empty_string()
        print(f'{_PASS} Check 4/{total}: test_empty_string() passes')
        score += 1
    except AssertionError as e:
        print(f'{_FAIL} Check 4/{total}: test_empty_string() failed — {e}')
    except Exception as e:
        print(f'{_FAIL} Check 4/{total}: unexpected error — {e}')

    print()
    if score == total:
        print('=' * 52)
        print(f'  {_PASS}  Exercise 3 complete! {total}/{total} checks passed.')
        print('=' * 52)
    else:
        print(f'  {score}/{total} passed. Keep going!')

_run_checks()
````

**Key details:** (1) Each test function has one `assert` and one case.

(3) A passing test returns `None` silently — no news is good news.
