## Bonus Challenge

Add a test that checks the error *message* too:

```python
def test_divide_error_message():
    with pytest.raises(ValueError, match="Cannot divide by zero"):
        divide(5, 0)
```

The `match` argument is a regex pattern. If the exception message doesn't
match, the test fails even though a `ValueError` was raised. This is useful
when a function raises the same exception type from multiple paths.

## Solution

<details>
<summary>Click to reveal — try on your own first</summary>

```python
import pytest

def test_divide_normal():
    assert divide(10, 2) == 5.0


def test_divide_raises():
    with pytest.raises(ValueError):
        divide(10, 0)
```

**Key details:** (1) `test_divide_normal` checks the happy path with a single
`assert`. (2) `test_divide_raises` uses the `with pytest.raises(ValueError):`
context manager. If `divide(10, 0)` does NOT raise `ValueError`, the test
fails. (3) The `import pytest` at the top of the setup cell makes `pytest`
available in the notebook namespace.
