```python
import logging

logging.basicConfig(
    level=logging.DEBUG,
    format="%(asctime)s %(levelname)-8s %(message)s",
)

logger = logging.getLogger(__name__)

def safe_parse(text: str) -> float | None:
    try:
        return float(text.strip())
    except ValueError as e:
        logger.warning("safe_parse() | could not parse %r: %s", text, e)
        return None
```

**Key details:** (1) Catch `ValueError` specifically, not bare `except`. (2) Log
at WARNING (not ERROR) — a parse failure is expected input, not a system failure.
(3) Return `None` as the safe fallback. (4) Use `%r` for `text` so the repr
(with quotes and whitespace visible) appears in the log.
