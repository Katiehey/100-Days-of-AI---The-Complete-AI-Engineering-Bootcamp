Write a second version using `.get('active', False)` so the function doesn't crash on a record that's missing the `active` key.

## Solution

<summary>Click to reveal — try on your own first</summary>

```python
def active_emails(users: list) -> list:
    return [u['email'].lower() for u in users if u['active']]
```

**Why this works:** The comprehension does two jobs at once. `for u in users` walks every record; `if u['active']` filters to just the active ones; and `u['email'].lower()` is the value we keep for each — the lowercased email. One line, one clear intent.
