```python
def clean_text(text: str) -> str:
    text = text.strip().lower()
    return ' '.join(text.split())
```

**Why this works:** `.strip()` and `.lower()` handle the ends and the casing. `text.split()` with no argument splits on *any* run of whitespace (spaces, tabs, newlines) and discards the empties; `' '.join(...)` reconnects the pieces with single spaces — collapsing all internal whitespace.
