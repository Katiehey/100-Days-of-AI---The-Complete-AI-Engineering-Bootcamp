Also catch `json.JSONDecodeError` in `load_json` so a corrupt file returns `{}` instead of crashing.

```python
import json

def save_json(data, path: str) -> None:
    with open(path, 'w') as f:
        json.dump(data, f, indent=2)

def load_json(path: str):
    try:
        with open(path) as f:
            return json.load(f)
    except FileNotFoundError:
        return {}
```

**Why this works:** `json.dump` writes the object to an open file handle; the `with` block closes it cleanly. `load_json` wraps the read in `try/except FileNotFoundError`, so a missing file returns a safe `{}` fallback instead of crashing your program.
