```python
def clean_record(record: dict) -> dict:
    name = ' '.join(record.get('name', '').split()).title()
    email = record.get('email', '').strip().lower()
    return {'name': name, 'email': email}
```

**Why this works:** `.get('name', '')` returns `''` when the key is missing, so nothing crashes. `' '.join(...split())` collapses any internal whitespace (the idiom from Lesson 2), `.title()` fixes the casing, and the email is stripped and lowercased — a fresh, fully-clean dict.
