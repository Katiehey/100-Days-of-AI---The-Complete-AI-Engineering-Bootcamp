```python
def normalize_name(name: str) -> str:
    return name.strip().title()
```

**Why this works:** `.strip()` removes surrounding whitespace and returns a new string; `.title()` capitalises the first letter of each word and lowercases the rest. Because each method returns a new string, we chain them left to right.
