Return a tuple `(text, model_name)` instead of just the text. Hint: `response["model"]` contains the model that ran. This is useful when you run multiple models and want to know which one answered.

```python
import ollama

MODEL = "llama3.2"

def ping_ollama(question: str) -> str:
    response = ollama.chat(
        model=MODEL,
        messages=[{"role": "user", "content": question}]
    )
    return response["message"]["content"]
```

**Why this works:** `ollama.chat()` connects to the local Ollama server (which must be running), sends the model name and messages list, waits for the response, and returns a Python dict. The reply text lives at `response["message"]["content"]` — nested because the response mirrors the same role/content structure as the request, just with role set to "assistant".
