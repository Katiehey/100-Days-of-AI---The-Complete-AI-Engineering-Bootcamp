```python
import ollama

MODEL = "llama3.2"

def ask_with_system(system_prompt: str, question: str) -> str:
    messages = [
        {"role": "system", "content": system_prompt},
        {"role": "user",   "content": question},
    ]
    response = ollama.chat(model=MODEL, messages=messages)
    return response["message"]["content"]
```

**Why this works:** The system message is the first entry in the messages list with `role` set to `"system"`. The model processes messages in order, so it reads the system instructions before the user question. Note this differs from some other APIs (like the Anthropic SDK) where the system prompt is a separate top-level parameter — here it lives inside the messages list as a special role.
