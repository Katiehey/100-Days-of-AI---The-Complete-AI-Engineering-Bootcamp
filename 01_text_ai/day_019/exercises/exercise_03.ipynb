# Day 019 — Exercise 3: TestCase and EvalResult

**Goal:** Implement the `TestCase` and `EvalResult` dataclasses. Use `field(default_factory=list)` for list fields — this prevents the classic shared-mutable-default bug.

````python
from dataclasses import dataclass, field
````

## Your Implementation

````python
@dataclass
class TestCase:
    """
    One test input with its expected outputs.

    Fields:
        question           -- the text sent to the model
        expected_keywords  -- keywords that should appear in any correct answer
        expected_answer    -- optional full answer for llm_judge scoring
    """
    question: str
    # TODO: expected_keywords: list[str] = field(default_factory=list)
    # TODO: expected_answer: str = ''


@dataclass
class EvalResult:
    """
    The outcome of running one TestCase through the model.

    Fields:
        test_case         -- the original TestCase
        response          -- the model's actual response text
        passed            -- True if keyword check passed
        matched_keywords  -- which keywords were found
        judge_score       -- LLM judge score 1-5 (0 = not run)
        judge_rationale   -- LLM judge explanation (empty = not run)
    """
    test_case: TestCase
    response: str
    passed: bool
    # TODO: matched_keywords: list[str] = field(default_factory=list)
    # TODO: judge_score: int = 0
    # TODO: judge_rationale: str = ''
````

## Check Your Work

````python
def _run_checks():
    total = 5
    passed = 0

    # Check 1: TestCase and EvalResult defined
    try:
        assert 'TestCase' in globals() and 'EvalResult' in globals()
        passed += 1; print('✅ Check 1: TestCase and EvalResult defined')
    except Exception as e:
        print(f'❌ Check 1: {e}')

    # Check 2: TestCase fields accessible
    try:
        tc = TestCase('What is the capital of France?',
                      expected_keywords=['paris'], expected_answer='Paris')
        assert tc.question == 'What is the capital of France?'
        assert tc.expected_keywords == ['paris']
        assert tc.expected_answer == 'Paris'
        passed += 1; print('✅ Check 2: TestCase fields accessible')
    except Exception as e:
        print(f'❌ Check 2: TestCase fields — {e}')

    # Check 3: EvalResult fields accessible
    try:
        tc2 = TestCase('Test?')
        er = EvalResult(test_case=tc2, response='Test response', passed=True)
        assert er.passed == True
        assert er.judge_score == 0
        assert er.judge_rationale == ''
        passed += 1; print('✅ Check 3: EvalResult fields accessible with defaults')
    except Exception as e:
        print(f'❌ Check 3: EvalResult fields — {e}')

    # Check 4: default expected_keywords is empty list
    try:
        tc3 = TestCase('No keywords?')
        assert tc3.expected_keywords == []
        assert tc3.expected_answer == ''
        passed += 1; print('✅ Check 4: TestCase defaults work (empty keywords, empty answer)')
    except Exception as e:
        print(f'❌ Check 4: defaults — {e}')

    # Check 5: two TestCases have INDEPENDENT keyword lists (mutable default bug check)
    try:
        t1 = TestCase('Q1?')
        t2 = TestCase('Q2?')
        t1.expected_keywords.append('keyword')
        assert t2.expected_keywords == [], \
            't2 keywords were modified by appending to t1 — use field(default_factory=list)'
        passed += 1; print('✅ Check 5: two TestCase instances have independent keyword lists')
    except Exception as e:
        print(f'❌ Check 5: mutable default — {e}')

    if passed == total:
        print('🎉 Exercise complete!')
    print(f'\nScore: {passed}/{total}')

_run_checks()
````

from dataclasses import dataclass, field

class TestCase:
    question: str
    expected_keywords: list[str] = field(default_factory=list)
    expected_answer: str = ''

@dataclass
class EvalResult:
    test_case: TestCase
    response: str
    passed: bool
    matched_keywords: list[str] = field(default_factory=list)
    judge_score: int = 0
    judge_rationale: str = ''
