```python
def make_system_prompt(role: str, constraints: list[str], output_style: str) -> str:
    constraint_lines = "\n".join(f"- {c}" for c in constraints)
    return (
        f"## Role\n{role}\n\n"
        f"## Constraints\n{constraint_lines}\n\n"
        f"## Output style\n{output_style}"
    )
```

**Why this works:** Each of the three arguments maps to a clearly labelled section in the final string. Using f-strings to join constraint items as a bullet list means the model sees a structured, scannable format rather than a wall of prose — large language models follow labelled sections more reliably than unlabelled paragraphs. Keeping the three parts separate in the function signature also means you can change tone, swap constraints, or update the output style independently without touching the other parts.
