You run this notebook cell-by-cell. It defines a **prompt library** with five

**Concepts composed from today's lessons:**

  (editor, teacher, analyst, etc.)
- **Lesson 3** — `extract_keywords` uses few-shot examples so the model

  shape (one paragraph, a bullet list, numbered keywords, one sentence, etc.)
- **Lesson 5** — everything is collected into a `PROMPT_LIBRARY` dict of named
  builder functions; `apply_transform(name, text)` is the single call site
