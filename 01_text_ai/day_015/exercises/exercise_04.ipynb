## Check Your Work

````python
def _run_checks():
    total = 5
    passed = 0

    system_msg = {"role": "system", "content": "You are a pirate."}

    # Check 1: returns a list
    try:
        h = [system_msg, {"role": "user", "content": "Ahoy"}]
        result = reset_history(h)
        assert isinstance(result, list), f"Expected list, got {type(result)}"
        passed += 1
        print("\u2705 Check 1: returns a list")
    except Exception as e:
        print(f"\u274c Check 1: returns a list \u2014 {e}")

    # Check 2: returns only system message when present
    try:
        h = [system_msg,
             {"role": "user",      "content": "Hello"},
             {"role": "assistant", "content": "Arr!"}]
        result = reset_history(h)
        assert len(result) == 1, f"Expected 1 message (system only), got {len(result)}"
        assert result[0]["role"] == "system", f"Expected system role, got {result[0]['role']!r}"
        passed += 1
        print("\u2705 Check 2: returns only the system message")
    except Exception as e:
        print(f"\u274c Check 2: system-only result \u2014 {e}")

    # Check 3: system message content is preserved exactly
    try:
        h = [system_msg, {"role": "user", "content": "Hi"}]
        result = reset_history(h)
        assert result[0]["content"] == system_msg["content"], \
            f"System content changed: {result[0]['content']!r}"
        passed += 1
        print("\u2705 Check 3: system message content is preserved exactly")
    except Exception as e:
        print(f"\u274c Check 3: content preservation \u2014 {e}")

    # Check 4: returns empty list when no system message
    try:
        h = [{"role": "user",      "content": "Hello"},
             {"role": "assistant", "content": "Hi"}]
        result = reset_history(h)
        assert result == [], f"Expected [], got {result}"
        passed += 1
        print("\u2705 Check 4: returns empty list when history has no system message")
    except Exception as e:
        print(f"\u274c Check 4: no-system case \u2014 {e}")

    # Check 5: original history is not mutated
    try:
        h = [system_msg,
             {"role": "user",      "content": "Q"},
             {"role": "assistant", "content": "A"}]
        original_len = len(h)
        _ = reset_history(h)
        assert len(h) == original_len, \
            f"Original mutated: was {original_len} msgs, now {len(h)}"
        passed += 1
        print("\u2705 Check 5: original history is not mutated")
    except Exception as e:
        print(f"\u274c Check 5: immutability \u2014 {e}")

    if passed == total:
        print("\U0001f389 Exercise complete!")
    print(f"\nScore: {passed}/{total}")

_run_checks()
````
