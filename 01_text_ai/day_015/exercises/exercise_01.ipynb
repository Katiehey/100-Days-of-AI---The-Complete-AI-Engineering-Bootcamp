def append_turn(
    history: list[dict],
    user_text: str,
    assistant_text: str,
) -> list[dict]:
    """Return a new history list with one user+assistant turn appended."""
    return history + [
        {"role": "user",      "content": user_text},
        {"role": "assistant", "content": assistant_text},
    ]
