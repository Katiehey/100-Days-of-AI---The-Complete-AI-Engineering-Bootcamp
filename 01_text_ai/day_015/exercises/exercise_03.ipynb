```python
def truncate_history(
    history: list[dict],
    max_turns: int = 10,
) -> list[dict]:
    """Keep the system prompt and the most recent max_turns user/assistant pairs."""
    if not history:
        return []
    if history[0]["role"] == "system":
        system = [history[0]]
        tail   = history[1:]
    else:
        system = []
        tail   = history
    keep = tail[-(max_turns * 2):]
    return system + keep
```
