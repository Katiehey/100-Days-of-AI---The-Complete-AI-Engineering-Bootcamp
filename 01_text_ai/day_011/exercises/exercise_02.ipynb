```python
import ollama


def get_embedding(text: str, model: str) -> list[float]:
    """Return the embedding vector for the given text using the specified model.

    Args:
        text:  The input string to embed.
        model: The Ollama embedding model to use (e.g. "nomic-embed-text").

    Returns:
        A plain Python list of floats representing the embedding vector.
        The length of the list is fixed by the model (768 for nomic-embed-text).

    Example:
        vector = get_embedding("The sky is blue.", "nomic-embed-text")
        print(len(vector))   # 768
        print(type(vector))  # <class 'list'>
    """
    response = ollama.embeddings(model=model, prompt=text)
    return response["embedding"]
```

**Why this works:** `ollama.embeddings()` takes the model name and the input text (called `prompt` in Ollama's API) and returns a plain Python dict. The embedding vector lives at the key `"embedding"` and is already a `list[float]` — no conversion needed. Passing `model` as a parameter rather than hard-coding it keeps the function reusable across different embedding models without any changes to the call sites.
