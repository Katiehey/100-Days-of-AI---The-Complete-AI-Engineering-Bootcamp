```python
import math


def cosine_similarity(a: list[float], b: list[float]) -> float:
    """Cosine similarity between two non-zero vectors."""
    dot = sum(ai * bi for ai, bi in zip(a, b))
    mag_a = math.sqrt(sum(x * x for x in a))
    mag_b = math.sqrt(sum(x * x for x in b))
    if mag_a == 0.0 or mag_b == 0.0:
        return 0.0
    return dot / (mag_a * mag_b)
```

**Why this works:** The dot product captures how much the two vectors agree element-by-element; dividing by each vector's magnitude removes the effect of length, leaving only the cosine of the angle between them. Checking for a zero magnitude before dividing prevents a `ZeroDivisionError` in the degenerate case where one vector is all zeros — returning `0.0` is a safe sentinel because a zero vector has no meaningful direction.
