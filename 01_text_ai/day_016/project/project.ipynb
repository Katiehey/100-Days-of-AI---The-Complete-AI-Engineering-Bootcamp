# Day 016 — Project: Streaming Chatbot

## Your Task

Build a streaming CLI chatbot with a **custom persona**.

Your bot must:

1. Use a custom `SYSTEM_PROMPT` that gives the bot a distinctive personality
   (a friendly tutor, a pirate, a Socratic philosopher — your choice)
2. Stream all replies token by token using `stream_chat_turn`
3. Support `/quit`, `/reset`, and `/history` commands
4. Truncate history at `max_turns=10` to keep the context window bounded
5. Handle `Ctrl+C` gracefully with a `try/except (EOFError, KeyboardInterrupt)`

## Deliverable

A `run_streaming_chatbot(system_prompt, model, max_turns)` function and a
custom `SYSTEM_PROMPT` constant. Run it in a terminal with
`run_streaming_chatbot()` — tokens should appear one by one as the model
generates them.

## Starter Hints

- Copy the five functions from today's exercises into a single script
  (or this notebook) and call `run_streaming_chatbot()` at the bottom
- The persona lives entirely in `SYSTEM_PROMPT` — you can change the
  bot's entire personality by changing one string
- Try asking a question that produces a long reply to see the streaming
  effect most clearly
