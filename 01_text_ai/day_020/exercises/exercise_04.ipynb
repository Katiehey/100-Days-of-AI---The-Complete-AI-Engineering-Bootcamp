```python
def build_cited_prompt(question: str, chunks: list[dict]) -> str:
    context = "\n\n".join(
        f"[{i+1}] Source: {c['source']}\n{c['text']}"
        for i, c in enumerate(chunks)
    )
    return f"Context:\n{context}\n\nQuestion: {question}"
```
