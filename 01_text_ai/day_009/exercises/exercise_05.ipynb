```python
def split_sections(text: str, section_headers: list[str]) -> dict[str, str]:
    """Split a multi-section document into labelled sections."""
    # Normalise header names for case-insensitive matching
    header_set = {h.strip().upper() for h in section_headers}

    # Accumulate lines per section; everything before the first header is
    # the preamble
    sections: dict[str, list[str]] = {"PREAMBLE": []}
    current = "PREAMBLE"

    for line in text.splitlines():
        normalised = line.strip().upper()
        if normalised in header_set:
            # This line is a header — switch the active section
            current = normalised
            sections.setdefault(current, [])
        else:
            # Regular content line — append to the current section
            sections.setdefault(current, []).append(line)

    # Join each section's lines into a single string and drop empty sections
    return {
        label: "\n".join(lines).strip()
        for label, lines in sections.items()
        if "\n".join(lines).strip()
    }
```

**Why this works:** The function tracks a `current` label and switches it whenever a line matches a known header (after stripping and upper-casing both sides so comparison is always case-insensitive). Lines that are not headers accumulate in a list under `current`, so each section's text is collected in one pass. The preamble key is initialised as `"PREAMBLE"` (uppercase), consistent with all other section keys being uppercase. Joining and stripping at the end converts the lists back to clean strings and silently drops any section that ended up empty — no special-casing required.
