## The Anti-Pattern — Do Not Edit This Cell

These three functions follow the prompt-as-string anti-pattern from Lesson 1: they embed variable data directly in f-strings and mix prompt construction with the model call. Your task is to refactor them — do not change this cell.

````python
import ollama

MODEL = "llama3.2"


# ── Anti-pattern functions (read-only — refactor below) ───────────────────────

def summarize(doc: str) -> str:
    """Summarize a document in one sentence."""
    prompt = (
        "Summarize the following text in one sentence.\n"
        "Be concise and factual.\n\n"
        f"{doc}"
    )
    response = ollama.chat(
        model=MODEL,
        messages=[{"role": "user", "content": prompt}],
    )
    return response["message"]["content"]


def classify(text: str, labels: list) -> str:
    """Classify text into one of the given labels."""
    label_str = ", ".join(labels)
    prompt = (
        f"Classify the text below as exactly one of: {label_str}.\n"
        "Reply with the label only — no punctuation, no explanation.\n\n"
        f"{text}"
    )
    response = ollama.chat(
        model=MODEL,
        messages=[{"role": "user", "content": prompt}],
    )
    return response["message"]["content"].strip()


def translate(text: str, target_language: str) -> str:
    """Translate text into the target language."""
    prompt = (
        f"Translate the following text into {target_language}.\n"
        "Preserve the original meaning. Reply with the translation only.\n\n"
        f"{text}"
    )
    response = ollama.chat(
        model=MODEL,
        messages=[{"role": "user", "content": prompt}],
    )
    return response["message"]["content"]
````

```python
import ollama

MODEL = "llama3.2"


# ── Template functions (pure — no ollama calls) ───────────────────────────────

def summarize_prompt(doc: str) -> str:
    return (
        "Summarize the following text in one sentence.\n"
        "Be concise and factual.\n\n"
        f"{doc}"
    )


def classify_prompt(text: str, labels: list) -> str:
    label_str = ", ".join(labels)
    return (
        f"Classify the text below as exactly one of: {label_str}.\n"
        "Reply with the label only — no punctuation, no explanation.\n\n"
        f"{text}"
    )


def translate_prompt(text: str, target_language: str) -> str:
    return (
        f"Translate the following text into {target_language}.\n"
        "Preserve the original meaning. Reply with the translation only.\n\n"
        f"{text}"
    )


# ── Caller functions ──────────────────────────────────────────────────────────

def summarize_v2(doc: str) -> str:
    prompt = summarize_prompt(doc)
    response = ollama.chat(
        model=MODEL,
        messages=[{"role": "user", "content": prompt}],
    )
    return response["message"]["content"]


def classify_v2(text: str, labels: list) -> str:
    prompt = classify_prompt(text, labels)
    response = ollama.chat(
        model=MODEL,
        messages=[{"role": "user", "content": prompt}],
    )
    return response["message"]["content"].strip()


def translate_v2(text: str, target_language: str) -> str:
    prompt = translate_prompt(text, target_language)
    response = ollama.chat(
        model=MODEL,
        messages=[{"role": "user", "content": prompt}],
    )
    return response["message"]["content"]


# ── Assertion tests (no model call) ──────────────────────────────────────────

doc = 'Photosynthesis converts sunlight into energy.'
assert doc in summarize_prompt(doc)
assert 'sentence' in summarize_prompt(doc).lower()

assert 'positive' in classify_prompt('great!', ['positive', 'negative'])
assert 'negative' in classify_prompt('great!', ['positive', 'negative'])

assert 'French' in translate_prompt('Hello', 'French')
assert 'Hello' in translate_prompt('Hello', 'French')
```

**Why this works:** Each `_prompt` function is a pure function — it takes data in, returns a string, and has no side effects. The caller functions own the model interaction and delegate prompt construction entirely to the template function. This seam between template and caller is what allows the assertion tests to verify prompt correctness without a network call.
