# Day 060 — Project: Caching & Performance

Build `caching_api.py` — a FastAPI server that caches Ollama responses to eliminate redundant model calls and measure cache effectiveness.

## Deliverable

`caching_api.py` in `project/solution/` — a FastAPI app with:

| Endpoint | Method | Description |
|----------|--------|-------------|
| `/health` | GET | Health check |
| `/ask` | POST | Cached Ollama Q&A (`cache_hit` flag in response) |
| `/cache/stats` | GET | Hits, misses, current cache size |
| `/cache` | DELETE | Clear cache and reset stats |

## How to run

```bash
CACHE_TTL_SECONDS=300 uvicorn caching_api:app --reload
# Ask the same question twice:
curl -X POST http://localhost:8000/ask \\
     -H 'Content-Type: application/json' \\
     -d '{"prompt": "What is Python?"}'
# Second call is instant (cache_hit: true):
curl -X POST http://localhost:8000/ask \\
     -H 'Content-Type: application/json' \\
     -d '{"prompt": "What is Python?"}'
```

## Concepts used

- `SimpleCache` with TTL — in-memory key-value store
- Cache-aside pattern — check cache, call fn on miss
- `cache_hit` flag — transparency about caching behaviour
- Stats endpoint — operational visibility into cache effectiveness
- `CACHE_TTL_SECONDS` env var — TTL configurable without code changes
