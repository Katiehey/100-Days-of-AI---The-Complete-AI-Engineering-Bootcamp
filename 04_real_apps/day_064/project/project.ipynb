# Day 064 — Project: Capstone Build I

Build `writing_assistant.py` — the MVP of an AI Writing Assistant product.

## What You're Building

`writing_assistant.py` — a FastAPI app that combines everything from Section 4:

| Feature | From Day |
|---------|----------|
| FastAPI routes | 52 |
| Pydantic request/response | 52 |
| In-process state (ContentStore) | 54 |
| Health endpoint | 57 |
| Feature gating | 63 |
| Rate limiting (429) | 63 |
| Template engine | 064 |
| Content history | 064 |

## Endpoints

```
GET  /health                  — health check
GET  /plan                    — plan, usage, limit
GET  /templates               — list template names
POST /generate                — free-text generation
POST /generate/template       — template-based generation
POST /improve                 — improve text (pro tier only)
GET  /history/{user_id}       — content history
GET  /content/{content_id}    — retrieve one item
```

## Run

```bash
uvicorn writing_assistant:app --reload
# → http://localhost:8000/docs
```
