# Day 064 — Exercise 1: MVP Scoping

**Scope creep** is the #1 reason MVPs never ship. The fix: write down every feature idea, then ruthlessly slot them into tiers:

| Tier | Rule | Goal |
|------|------|------|
| Free | First N features | Attract users, demonstrate value |
| Pro | Next M features | Justify upgrade |
| Backlog | Everything else | Build *after* first paying customer |

A good MVP has 2-3 free features and 2-3 pro features. Everything else waits until the product is validated.

## Task

Implement `plan_mvp(features, free_count, pro_count) -> dict`:

- `free_tier`: first `free_count` features from the list
- `pro_tier`: the next `pro_count` features
- `backlog`: all remaining features
- `summary`: `'{n} free / {m} pro / {k} backlog'`
- If fewer features than `free_count`: all go to `free_tier`, others empty

````python
def plan_mvp(features: list[str], free_count: int,
             pro_count: int) -> dict:
    free    = features[:free_count]
    pro     = features[free_count: free_count + pro_count]
    backlog = features[free_count + pro_count:]
    return {
        "free_tier": free,
        "pro_tier":  pro,
        "backlog":   backlog,
        "summary":   f"{len(free)} free / {len(pro)} pro / {len(backlog)} backlog",
    }
````

**Three list slices**: `[:free_count]`, `[free_count:free_count+pro_count]`, `[free_count+pro_count:]`. Slicing past the end of a list returns an empty list — no bounds checks needed. The summary string is an f-string using `len()` of each slice.
