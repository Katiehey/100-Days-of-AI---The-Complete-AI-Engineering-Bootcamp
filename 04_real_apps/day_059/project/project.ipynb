# Day 059 — Project: Background Job Processing API

Build `background_api.py` — a FastAPI server that accepts long-running Ollama summarization requests, processes them in background threads, and lets clients poll for results.

## Deliverable

`background_api.py` in `project/solution/` — a FastAPI app with:

| Endpoint | Method | Description |
|----------|--------|-------------|
| `/health` | GET | Health check |
| `/summarize` | POST | Submit text → background Ollama summarization |
| `/jobs/{job_id}` | GET | Poll status / retrieve result |

## How to run

```bash
uvicorn background_api:app --reload
# Then in another terminal:
curl -X POST http://localhost:8000/summarize \\
     -H 'Content-Type: application/json' \\
     -d '{"text": "Long document text here..."}'
# Returns: {"job_id": "abc123", "status": "pending"}
curl http://localhost:8000/jobs/abc123
# Returns: {"status": "done", "result": "One-sentence summary."}
```

## Concepts used

- `_JobStore` — thread-safe status/result store with `threading.Lock`
- `threading.Thread(target=worker, daemon=True).start()` — background thread
- `daemon=True` — thread exits when main process exits (no zombie threads)
- `build_api(process_fn=None)` — dependency injection for testability
- `POST /summarize → 202 Accepted` then `GET /jobs/{id}` polling pattern
