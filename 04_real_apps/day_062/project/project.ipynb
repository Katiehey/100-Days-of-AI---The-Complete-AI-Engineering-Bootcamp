# Day 062 — Project: Testing Web Apps

Build `test_app.py` — a real pytest test suite for a CRUD API. When complete, run it with `pytest test_app.py -v` and see all tests pass.

## Deliverable

`test_app.py` in `project/solution/` — a pytest test suite with:

| Feature | Detail |
|---------|--------|
| `@pytest.fixture` | Fresh TestClient for every test — no shared state |
| `@pytest.mark.parametrize` | Test many input combinations in one test |
| Happy path tests | 200/201/204 for valid requests |
| Validation tests | 422 for invalid payloads |
| Not found tests | 404 for missing resources |
| State transition | Create → GET → DELETE → 404 |

## How to run

```bash
cd 04_real_apps/day_062/project/solution/
pytest test_app.py -v          # run all tests, verbose
pytest test_app.py -k delete   # run only delete tests
pytest test_app.py --tb=short  # compact tracebacks
```

## Concepts used

- `@pytest.fixture` — shared setup (TestClient) reset per test
- `@pytest.mark.parametrize` — table-driven testing
- `assert r.status_code == N` — clear status assertions
- State transition tests — POST → DELETE → 404
- Test isolation — `build_app()` called fresh in the fixture
