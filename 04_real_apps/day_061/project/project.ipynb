# Day 061 — Project: Monitoring & Logging

Build `monitored_api.py` — a FastAPI server that logs every request as structured JSON, tracks metrics, and exposes health and metrics endpoints.

## Deliverable

`monitored_api.py` in `project/solution/` — a FastAPI app with:

| Endpoint | Method | Description |
|----------|--------|-------------|
| `/health` | GET | Health check |
| `/metrics` | GET | Request count, errors, avg latency, error rate |
| `/ask` | POST | Ollama Q&A (JSON-logged) |

**Features:**
- `JsonFormatter` — every log line is a parseable JSON object
- `MetricsCollector` middleware — records status + duration for every request
- `LOG_LEVEL` env var — controls verbosity without code changes

## How to run

```bash
LOG_LEVEL=DEBUG uvicorn monitored_api:app --reload
# Watch structured JSON logs appear per request:
curl -X POST http://localhost:8000/ask \\
     -H 'Content-Type: application/json' \\
     -d '{"prompt": "What is observability?"}'
# Check metrics:
curl http://localhost:8000/metrics
```

## Concepts used

- `logging.getLogger`, `setLevel`, `StreamHandler` — stdlib logging setup
- `JsonFormatter` — one JSON object per log line
- `MetricsCollector` — request count, errors, latency
- `@app.middleware('http')` — intercept every request/response
- `time.monotonic()` for latency measurement
