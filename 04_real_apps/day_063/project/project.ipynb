# Day 063 — Project: Payments & Productization

Build `gated_api.py` — a FastAPI server with plan-based feature gating, rate limiting, and Stripe Checkout integration.

## Deliverable

`gated_api.py` in `project/solution/` — a FastAPI app with:

| Endpoint | Description |
|----------|-------------|
| `GET /health` | Health check |
| `GET /plan` | Current plan, usage, limit |
| `POST /ask` | Rate-limited AI Q&A |
| `POST /checkout` | Create Stripe checkout session |
| `GET /checkout/success` | Handle post-payment redirect |
| `GET /checkout/cancel` | Handle cancelled checkout |
| `POST /webhook` | Handle Stripe webhook events |

## Setup (requires a free Stripe account)

```bash
pip install stripe
export STRIPE_SECRET_KEY=sk_test_...  # from dashboard.stripe.com
export STRIPE_WEBHOOK_SECRET=whsec_...
uvicorn gated_api:app --reload
```

## Testing without Stripe credentials

The `build_api(process_fn=..., initial_plan=...)` factory accepts a `process_fn` for local testing without Ollama or Stripe.

## Concepts used

- Feature matrix: plan → set of allowed features
- Rate limiting: usage_count >= daily_limit → 429
- Stripe Checkout: session create → redirect URL
- Webhook events: subscription.created/deleted → plan change
- `stripe.Webhook.construct_event` — signature verification
