## Solution

<details><summary>Reveal</summary>

```python
def check_feature_access(plan: str, feature: str) -> bool:
    return feature in FEATURE_MATRIX.get(plan, set())
```

**One line thanks to `dict.get(key, default)`**: if the plan is unknown, `get` returns an empty set, so `feature in set()` is always `False` — safe by default. Adding a new plan is a single dict entry; no code changes.
