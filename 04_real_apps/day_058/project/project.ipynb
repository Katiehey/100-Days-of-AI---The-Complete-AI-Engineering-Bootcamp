# Day 058 — Project: Streaming Chat API

Build `streaming_api.py` — a production-ready FastAPI server that streams Ollama responses via **Server-Sent Events** (SSE) and **WebSockets**.

## Deliverable

`streaming_api.py` in `project/solution/` — a FastAPI app with four endpoints:

| Endpoint | Method | Description |
|----------|--------|-------------|
| `/health` | GET | Health check |
| `/stream/count` | GET | Demo SSE counting stream (no Ollama) |
| `/chat/stream` | POST | Streams Ollama tokens as SSE events |
| `/ws` | WebSocket | Bidirectional streaming chat |

## How to run

```bash
uvicorn streaming_api:app --reload
```

- REST docs: http://localhost:8000/docs
- Test SSE: `curl -N http://localhost:8000/stream/count?n=5`
- Test WebSocket: `wscat -c ws://localhost:8000/ws`

## Concepts used

- `StreamingResponse` + `media_type='text/event-stream'` → SSE
- Sync generator functions as the streaming body
- `ollama.chat(stream=True)` → token-by-token Ollama responses
- `WebSocket` / `WebSocketDisconnect` from FastAPI
- `ChatHistory` class for multi-turn context management
